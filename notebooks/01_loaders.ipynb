```python
import pandas as pd

# Caminho para o ficheiro com administrações inválidas.
invalid_administrations_path = (
    PROJECT_ROOT
    / "data"
    / "canonical"
    / "invalid"
    / "emergency_administrations.csv"
)

# Carregamos as administrações inválidas.
# A coluna administered_at é convertida logo para datetime.
df_invalid_administrations = pd.read_csv(
    invalid_administrations_path,
    parse_dates=["administered_at"]
)

# Mostramos os dados para confirmar que foram carregados.
df_invalid_administrations
```

```python
# Aplica a função que valida se cada administração
# referencia uma prescrição existente.
#
# Primeiro argumento:
# DataFrame com as administrações inválidas que queremos testar.
#
# Segundo argumento:
# DataFrame com as prescrições válidas existentes no sistema.
invalid_results = validate_prescription_exists(
    df_invalid_administrations,
    dataset["emergency_prescriptions"]
)

# Mostra o resultado produzido pelo validator.
invalid_results
```

```python
# Importamos a função responsável por validar
# se os códigos de medicamento existem no mapping.
from src.validators.medication_validator import (
    validate_medication_code_exists
)

# Validamos as administrações válidas usando
# a tabela de mapping de medicamentos do dataset.
valid_medication_results = validate_medication_code_exists(
    dataset["emergency_administrations"],
    dataset["medication_code_mapping"]
)

# Mostramos o resultado.
valid_medication_results
```

```python
# Aplicamos a mesma regra aos casos de teste inválidos.
invalid_medication_results = validate_medication_code_exists(
    df_invalid_administrations,
    dataset["medication_code_mapping"]
)

# Mostramos os resultados.
invalid_medication_results

```

```python
# Importamos a função que valida a coerência temporal
# entre prescrição e administração.
from src.validators.timing_validator import (
    validate_administration_after_prescription
)

# Validamos a administração correta.
valid_timing_results = validate_administration_after_prescription(
    dataset["emergency_administrations"],
    dataset["emergency_prescriptions"]
)

# Mostramos o resultado.
valid_timing_results
```

```python
# Importamos a função que junta todas as regras
# de validação das administrações.
from src.validators.validation_pipeline import (
    validate_emergency_administrations
)

# Executamos todas as regras de validação
# sobre a administração válida.
valid_report = validate_emergency_administrations(
    dataset["emergency_administrations"],
    dataset["emergency_prescriptions"],
    dataset["medication_code_mapping"]
)

# Mostramos o relatório consolidado.
valid_report
```
